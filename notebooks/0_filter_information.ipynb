# Read and Filter Information

```python
from pathlib import Path
import sys
import os
import json
sys.path.append(os.path.abspath('..'))
```

```python
with open('../config/target_person.json', 'r', encoding='utf8') as f:
    target_person = json.load(f)
```

```python
with open('../config/topology.json', 'r') as f:
    topology = json.load(f)
```

```python
from src.utils import load_keypoint_data_by_id
directory = Path('../dataset/raw/')
samples = []
for file_path in directory.glob('*.h5'):
    
    video_source = file_path.stem.lower().split('_')[0]
    person_idx = target_person[video_source]
    keypoints = load_keypoint_data_by_id(file_path, person_idx)
    
    samples.append(
        {
            "keypoints": keypoints,
            "label": 1 if video_source.find('suspicious') else 0,
            "topology": topology,
            "source": video_source
        }
    )
```

```python
from src.utils import save_dataset
save_dataset('../dataset/filtred', samples)
```
