# Read and Filter Information

```python
from pathlib import Path
import sys
import os
import json
sys.path.append(os.path.abspath('..'))
```

```python
with open('../config/target_person.json', 'r', encoding='utf8') as f:
    target_person = json.load(f)
```

```python
with open('../config/topology.json', 'r') as f:
    topology = json.load(f)
```

```python
from src.utils import load_keypoints_data_by_id, load_bbox_by_id
directory = Path('../dataset/raw/')
samples = []
for file_path in directory.glob('*.h5'):
    
    video_source = file_path.stem.lower().split('_')[0]
    person_idx = target_person[video_source]
    keypoints = load_keypoints_data_by_id(file_path, person_idx)
    bbox = load_bbox_by_id(file_path, person_idx)
    samples.append(
        {
            "keypoints": keypoints,
            "label": 1 if 'suspicious' in video_source else 0,
            "topology": topology,
            "source": video_source,
            "bbox": bbox
        }
    )

sources = [sample['source'] for sample in samples]
labels = [sample['label'] for sample in samples]
```

```python
dataset = {
    'samples': samples,
    'sources': sources,
    'labels': labels
}
```

```python
from src.utils import save_dataset
save_dataset('../dataset/filtred', dataset)
```
