# Stadistics Approach

```python

def compute_timeserie_statistics(sample: np.ndarray, funcs=[np.mean, np.std, np.min, np.max, np.median]):
    resultados = [f(sample, axis=0) for f in funcs]
    return np.concatenate(resultados).flatten()
```

```python
samples = [
    np.concatenate(([
        sample['label']], 
        compute_timeserie_statistics(sample['handcrafted']),
        [sample['source']]
    )) 
    for sample in dataset
]
```

```python
samples = np.array(samples)
```

```python
import pandas as pd
df = pd.DataFrame(samples)
df.to_csv('../dataset/features_statistics/dataset.csv', index=False, header=False)
```
